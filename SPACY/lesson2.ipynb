#### <b>Part Of Speech Tagging</b>
- In the field of computational linguistics, understanding parts-of-speech is essential. SpaCy offers an easy way to parse a text and identify its parts of speech. 
- Below, we will iterate across each token (word or punctuation) in the text and identify its part of speech.

```python
import spacy
text="Mike enjoys playing football ."
nlp=spacy.load("en_core_web_sm")
doc=nlp(text)
```

```python
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

```python
#Visualizing 
from spacy import displacy
displacy.render(doc,style="dep")
```

#### <b>Named Entity Recognition</b>
- Another essential task of NLP, is named entity recognition, or NER. Here, I’d like to demonstrate how to perform basic NER via spaCy.
- Again, we will iterate over the doc object as we did above, but instead of iterating over doc.sents, we will iterate over doc.ents. 
- For our purposes right now, I simply want to print off each entity’s text (the string itself) and its corresponding label (note the _ after label). 


```python
with open("wiki_us.txt","r") as f:
    text = f.read()
doc2=nlp(text)    
```

```python
for ent in doc2.ents:
    print(ent.text, ent.label_)#what label corresponds to what entity

#This model does some good job in NER but also does some mistakes tus we can opt for a larger model    
```

```python
from spacy import displacy
displacy.render(doc2,style="ent")
#This is an output where entities are highlighted
```
