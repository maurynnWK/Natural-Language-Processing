## Spacy Linguistic Annotations

```python
with open("wiki_us.txt",'r') as f:
    text=f.read()
print(text)   
```

```python
#With the data loaded in, it’s time to make our first Doc container
doc=nlp(text)
doc
```

- If you are trying to spot the difference between this and the text above, good luck. You will not see a difference when printing off the doc container. 
- But I promise you, it is quite different behind the scenes. The Doc container, unlike the text object, contains a lot of valuable metadata, or attributes, hidden behind it. 
- To prove this, let’s examine the length of the doc object and the text object.

```python
print("Length of text:",len(text))
print("Length of doc:",len(doc))
```

```python
#What’s going on here? Same text, but different length. Why does this occur?
#  To answer that, let’s explore it more deeply and try and print off each item in each object.
for token in text[0:30]:
    print(token)
```

```python
for token in doc[0:10]:
    print(token)

#More valuable    
```

- And now we see the magical difference. While on the surface it may seem that the Doc container’s length is dependent on the quantity of words, look more closely. 
- You should notice that the open and close parentheses are also considered an item in the container. These are all known as tokens. Tokens are a fundamental building block of spaCy or any NLP framework. 
- They can be words or punctuation marks. Tokens are something that has syntactic purpose in a sentence and is self-contained.

```python
#On the surface, this may not seem exceptional. But it is.
#You may be thinking to yourself that you could easily use 
#the split method in Python to split by whitespace and have the 
#same result. But you’d be wrong.

for token in text.split()[0:10] :
    print(token)
```

```python
#Notice that the parentheses are not removed or handled individually.
#To see this more clearly, let’s print off all tokens from index 
#5 to 8 in both the text and doc objects.
words=text.split()[0:10]
i=5
for token in doc[i:8]:
    print (f"SpaCy Token {i}:{token}\nWord Split {i}:{words[i]}\n\n")
    i=i+1

#The code compares how spaCy tokenizes text vs. how Python's split() method works.

#It highlights differences in tokenization, especially around punctuation, parentheses, or special characters.
#The comment in the code suggests that parentheses are not removed or handled individually, 
# meaning spaCy might treat them differently than split()    
```

#### <b>Sentence Boundary Detection</b>
- In NLP, sentence boundary detection, or SBD, is the identification of sentences in a text. Again, this may seem fairly easy to do with rules. One could use split(“.”), but in English we use the period to also denote abbreviation.
- You could, again, write rules to look for periods not proceeded by a lowercase word, but again, I ask the question, “why bother?”.
- We can use spaCy and in seconds have all sentences fully separated through SBD.

```python
for sent in doc.sents:#attribute of doc
    print(sent)
```

```python
#accessing a sentence
sentence1=doc.sent[0]
sentence1

# That is because the sents attribute is a generator.
# In python, we can usually iterate over generators by converting them into a list.
```

```python
sent1=list(doc.sents)[0]
sent1
```
